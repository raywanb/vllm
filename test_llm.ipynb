```python
from vllm import LLM
import os
os.environ["CUDA_VISIBLE_DEVICES"] = "4"
```

```python
llm = LLM(model="/data/raywanb/checkpoint/qwen3-swiftkv-8b-6-consumers-no-chat-template-every-4-layers/global_step_1501")
```

```python
from vllm import SamplingParams
```

```python
sampling_params = SamplingParams(
    temperature=0.0,
    max_tokens=1024,
    stop_token_ids=[151645, 151643]
)
```

```python
output = llm.generate("You are a helpful assistant. Please answer the following question: who is sam altman?", sampling_params=sampling_params)
```
